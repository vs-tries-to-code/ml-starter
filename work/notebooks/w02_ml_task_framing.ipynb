## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

**Refresh / Content Opportunity Scoring**
In order to prioritize pages based on whether they require refreshing or not, it requires a hybrid approach of both ranking and scoring.

The pages are to be scored based on their probability of declined and ranked in the order in which they need to be worked on.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

If using the starter dataset, it's a **proxy label** derived from a defined rule: ```trend_direction == "down"```.

If building a more advanced capstone, it's an **observed future outcome**: looking at the past 90 days of features to predict if traffic actually declined or recovered over the next 30 days.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Precision@50 would be a useful metric to measure success.

Editors have finite time and can only fix a finite number of candidates. Among them, the number of false positives are to be reduced for efficiency. Global metrics like ROC-AUC or standard accuracy evaluate the entire dataset, which doesn't reflect how a ranked list is consumed.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Real-world search behavior involves dozens of non-linear signals interacting at once—like the subtle trade-offs between content age, CTR, position tiers, engagement rates, and historical trend direction. A fixed if-else rule forces rigid, black-and-white cutoffs that miss nuanced combinations of decline. A machine learning model (like a random forest) can weigh all these continuous features simultaneously to catch messy, multi-signal patterns that a human rule would completely overlook.
